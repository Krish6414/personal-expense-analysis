```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)

# ---------------------------------------------------
# BASIC SETTINGS
# ---------------------------------------------------

num_transactions = 500

monthly_income = {
    "2026-01": 55000,
    "2026-02": 55000,
    "2026-03": 56000,
    "2026-04": 56000,
    "2026-05": 58000,
    "2026-06": 58000,
    "2026-07": 60000,
    "2026-08": 60000
}

# Monthly expense budgets
monthly_budget = {
    "2026-01": 36000,
    "2026-02": 34000,
    "2026-03": 37000,
    "2026-04": 35000,
    "2026-05": 39000,
    "2026-06": 36000,
    "2026-07": 40000,
    "2026-08": 38000
}

categories = {
    "Food": ["Restaurant", "Groceries", "Coffee", "Food Delivery"],
    "Transport": ["Uber", "Bus", "Auto", "Fuel", "Metro"],
    "Shopping": ["Clothes", "Shoes", "Electronics", "Accessories"],
    "Bills": ["Electricity", "Internet", "Mobile Bill", "Water Bill"],
    "Entertainment": ["Movies", "Games", "OTT Subscription", "Sports"],
    "Healthcare": ["Medicine", "Doctor Visit", "Pharmacy"],
    "Education": ["Books", "Online Course", "Study Material"],
    "Rent": ["House Rent"],
    "Travel": ["Flight", "Hotel", "Train", "Vacation"],
    "Other": ["Gift", "Donation", "Miscellaneous"]
}

payment_methods = [
    "UPI",
    "Credit Card",
    "Debit Card",
    "Cash",
    "Bank Transfer"
]

# Probability of selecting each expense category
category_weights = {
    "Food": 0.22,
    "Transport": 0.12,
    "Shopping": 0.10,
    "Bills": 0.10,
    "Entertainment": 0.08,
    "Healthcare": 0.05,
    "Education": 0.05,
    "Rent": 0.18,
    "Travel": 0.03,
    "Other": 0.07
}

transactions = []

transaction_id = 1

# ---------------------------------------------------
# GENERATE MONTHLY DATA
# ---------------------------------------------------

for month, income in monthly_income.items():

    # -----------------------------
    # Income transaction
    # -----------------------------

    year = int(month[:4])
    month_number = int(month[5:])

    salary_date = datetime(year, month_number, 1)

    transactions.append([
        f"TXN{transaction_id:04d}",
        salary_date,
        "Salary",
        "Monthly Salary",
        "Bank Transfer",
        income,
        "Income"
    ])

    transaction_id += 1

    # -----------------------------
    # Number of expenses this month
    # -----------------------------

    expense_count = random.randint(55, 65)

    budget = monthly_budget[month]

    # Generate random weights
    weights = np.random.dirichlet(
        np.ones(expense_count)
    )

    # Convert weights into expense amounts
    amounts = weights * budget

    # Minimum transaction amount
    amounts = np.maximum(amounts, 100)

    # Scale again to exactly match budget
    amounts = amounts / amounts.sum() * budget

    # -----------------------------
    # Generate expenses
    # -----------------------------

    for amount in amounts:

        category = random.choices(
            list(category_weights.keys()),
            weights=list(category_weights.values())
        )[0]

        description = random.choice(
            categories[category]
        )

        payment_method = random.choice(
            payment_methods
        )

        # Random day within the month
        if month_number == 12:
            next_month = datetime(year + 1, 1, 1)
        elif month_number == 8:
            next_month = datetime(year, 9, 1)
        else:
            next_month = datetime(year, month_number + 1, 1)

        days_in_month = (
            next_month - datetime(year, month_number, 1)
        ).days

        day = random.randint(1, days_in_month)

        date = datetime(
            year,
            month_number,
            day
        )

        transactions.append([
            f"TXN{transaction_id:04d}",
            date,
            category,
            description,
            payment_method,
            round(amount, 2),
            "Expense"
        ])

        transaction_id += 1


# ---------------------------------------------------
# CREATE DATAFRAME
# ---------------------------------------------------

df = pd.DataFrame(
    transactions,
    columns=[
        "Transaction_ID",
        "Date",
        "Category",
        "Description",
        "Payment_Method",
        "Amount",
        "Type"
    ]
)

# Sort by date
df = df.sort_values("Date").reset_index(drop=True)

# Month column
df["Month"] = df["Date"].dt.strftime("%Y-%m")

print("Dataset created successfully!")
print("Number of transactions:", len(df))
```
